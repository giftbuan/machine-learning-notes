```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Data Setup: Creating a dictionary of your backpacking costs
countries = ['Vietnam (Hanoi)', 'Laos', 'Cambodia', 'Vietnam (HCM)']
food = [3000, 10000, 7500, 7500]
transpo = [3500, 5500, 3000, 1500]
activities = [1000, 1000, 1000, 1000]

data = {
    'Destination': countries,
    'Food': food,
    'Transportation': transpo,
    'Activities': activities,
}

# 2. DataFrame: Converting the dictionary into a tabular format (Pandas Table)
df = pd.DataFrame(data)

# 3. Visualization: 'plt.figure' sets the canvas size
plt.figure(figsize=(12, 6))

# Plotting with styles: 'g-' (green line), marker="o" (circle points), lw=2 (line width)
plt.plot(df['Destination'], df['Food'], 'g-', marker="o", label='Food', mfc='white', lw=2)
plt.plot(df['Destination'], df['Transportation'], 'r-', marker='*', label='Transportation', mfc='white', lw=2)
plt.plot(df['Destination'], df['Activities'], 'b-', marker='x', label='Activities', mfc='white', lw=2)

# 4. Customization: Titles and axis labels make the plot readable
plt.title('Backpacking Budget Estimate by Country', fontsize=15)
plt.xlabel('Countries')
plt.ylabel('Estimate Cost (PHP)')
plt.grid(True, linestyle=':', alpha=0.6) # Grid helps in pinpointing values
plt.legend(loc='best') # 'best' automatically finds the clearest spot for labels
plt.yticks(range(0, 11001, 1000)) # Customizing Y-axis steps

plt.tight_layout() # Prevents label overlapping
plt.show()
```
