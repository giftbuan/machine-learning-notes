```python
import matplotlib.pyplot as plt

# Data: Comparing item prices across different Vietnamese stores (Convenience vs Market)
items = ['Bottled Water', 'Snacks/Chips', 'Soap/Shampoo', 'Instant Coffee']
convenience_store = [15, 45, 120, 80]
local_market = [10, 30, 90, 50]

# Bar Plot: Great for comparing discrete categories
plt.bar(items, convenience_store, color='orange', label='Convenience Store')
plt.bar(items, local_market, color='teal', label='Local Market', alpha=0.7)

plt.title('Vietnam Essential Item Price Comparison')
plt.ylabel('Cost (PHP)')
plt.legend()
plt.show()
```
